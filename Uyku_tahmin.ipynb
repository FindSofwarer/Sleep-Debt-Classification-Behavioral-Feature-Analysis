```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,classification_report
```

```python
df = pd.read_csv('bedtime_screentime_sleep_debt.csv')
df.head()
```

```python
df['phone_to_latency_ratio'] = df['bedtime_phone_minutes'] / (df['sleep_latency_min'] + 1)

# 2. Toplam verimli uyku oranı (Derin uyku + REM)
df['quality_sleep_pct'] = df['deep_sleep_pct'] + df['rem_sleep_pct']

# 3. Yorgunluk ve Kafein Etkileşimi
df['caffeine_per_activity'] = df['caffeine_post_5pm_mg'] / (df['physical_activity_min'] + 1)
```

```python
x = df.drop(columns=['user_id','sleep_debt_category'])
y = df['sleep_debt_category']
```

```python
x= pd.get_dummies(x,drop_first=True)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=30)
```

```python
for depth in [2,3,4,6,8,10]:
    temp_m = DecisionTreeClassifier(max_depth=depth,random_state=30)
    temp_m.fit(x_train,y_train)
    m_score = temp_m.score(x_test,y_test)
    print(f"max_depth = {depth} / score={m_score*100:.2f}")
```

```python
donusen_sutunlar = x.columns
ondem_dereceleri = temp_m.feature_importances_
df_önem = pd.DataFrame({'Sütun': donusen_sutunlar, 'Önem': ondem_dereceleri})
df_önem = df_önem.sort_values(by='Önem', ascending=False)

print("--- AĞACIN EN ÇOK GÜVENDİĞİ İLK 5 SÜTUN ---")
print(df_önem.head(5))
```
